```python
import pandas as pd
import numpy as np
```

missing values ko fill karrna dublicate ko remova karan


```python
from IPython.core.interactiveshell import default_banner
#sample daatset with missing values and dublicate

df= pd.DataFrame({
    'Name':['Alice','Bob','Charlie','David','Eve'],
    'Age':[25,30,np.nan,22,35],
    'Salary':[50000,60000,55000,np.nan,70000],})
df
#remove dublicate rows
df = df.drop_duplicates()
df
#filling missing valuse
df['Age'].fillna(df['Age'].mean(),inplace=True)
df['Salary'].fillna(df['Salary'].median(),inplace=True)
df
```
